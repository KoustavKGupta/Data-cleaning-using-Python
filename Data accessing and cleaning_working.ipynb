This is a data of 500 patients of which 350 patients participated in a clinical trial. None of them were using Novodra (a popular injectable insulin) or
Auralin (an oral insulin being researched) as a primary source of their insulin. All were experiencing elevated Hba1c level.

All 350 patients were treated with Novodra to establish a baseline Hba1c level or insulin dose. After 4 weeks which is not enough time to observe the change in Hba1c level that can be attributed by the switch to Novodra or Auralin.

1. 175 patients were switched to Auralin for 24 weeks
2. 175 patients were switched to Novodra for 24 weeks

Data for patient experiencing adverse effects is also recorded.

# Column Description
# Patients
1. Patient_id = Patient_id is the unique identifier for each patient in the master patient index (i.e. patient database) of the pharmaceutical company that is producing Auralin
2. The assigned_sex of each patient at birth (Male or Female)
3. given_name = the given name (i.e. first name) of each patient
4. surname = the surname (i.e. last name) of each patient
5. address = the main address for each patient
6. city = the city corresponding to the main address for each patient
7. state = the state corresponding to the main address for each patient
8. zip_code = the zip code corresponding to the main address for each patient
9. country = the country corresponding to the main address for each patient (All United States for this clinical trial)
10. contact = the phone number and email id for each patient
11. birthdate = the birthdate of each patient (mm/dd/yyyy). The inclusion criteria is age >= 18. There is no maximum age because diatebetes is a growing problem among elderly population
12. Weight = the weight of each patient in pounds (lbs).
13. Height = the height of each patient in inches.
14. bmi = the bmi index is body-mass index calculated by simply using a person's height and weight. The formula is kg/m2, where kg is the person's weight in kilogram and m2 is a person's height in metered square. A BMI of 25.0 or more is overweight, while the healthy range is 18.5 to 24.9. The inclusion criteria for this clinical trial is 16 >= bmi >= 38.

# Column description of treatments and treatments_cut
1. given_name = the given name of the patient in the master patient index who participated in the clinical trial
2. surname = the given name of the patient in the master patient index who participated in the clinical trial
3. Auralin = the baseline median of insulin dose from the week prior to switching to Auralin (the number before the dash) and the baseline insulin dose at the end of the treatment over the 24 weeks of treatment. Both are measured in units(u), which is international unit of measurement and standard measurement for insulin.
4. Novodra - Same as Auralin except for the patient who continued with Novodra.
5. hba1c_start - The patient's hba1c level at the beginning of the first week of the treatment. hba1c stands for hemoglobin a1c level. The hba1c test measures what the average blood sugar level has been over the last 3 or 4 months. It is a powerful way to get a overall sense of how blood sugar has been controlled. Everyone with diabetes should have this test 3 or 4 times in a year. Measured in %.
6. hba1c_end - The patients's blood sugar level at the end of the treatment.
7. hba1c_change - The change in patient's hba1c level from the start of the treatment to the end. For Auralin to be deemed effective, it must be non-inferior to Novodra, the current insulin standard. This non-inferiority is statistically defined as the upper bound of the 95% of the confidence interval being less than 0.4% for the difference between the mean changes of hba1c level for Auralin and Novodra (i.e. Novodra minus Auralin).

```python
# Zip_code is in string but decimal format. Need to remove decimal part and fill zeros upfrom to make it in 5 digit format.
# Function to make zip_code in 5 digit format
def clean_zip_code(zip_code):
    if zip_code == 'No Data':
        return 'NA'
    zip_code = int(float(zip_code)) # remove decimal part
    return str(zip_code).zfill(5)   # convert to string and pad with zero

patients_df['zip_code'].apply(clean_zip_code).str.len().value_counts()
```
